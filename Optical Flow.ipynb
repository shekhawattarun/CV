```python
# Optical Flow Computation and Visualization
# Load video
cap = cv2.VideoCapture('/content/sample_data/video.mp4')

# Parameters for Farneback Optical Flow
farneback_params = dict(
    pyr_scale=0.5, levels=3, winsize=15, iterations=3, poly_n=5, poly_sigma=1.2, flags=0
)

# Define video writer to save optical flow dance
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))
output = cv2.VideoWriter('optical_flow_dance.avi', cv2.VideoWriter_fourcc(*'XVID'), fps, (frame_width, frame_height))

# Read the first frame
ret, first_frame = cap.read()
gray1 = cv2.cvtColor(first_frame, cv2.COLOR_BGR2GRAY)

frame_count = 0  # Counter for saving frames as images

while True:
    ret, frame2 = cap.read()
    if not ret:
        break

    # Convert to grayscale
    gray2 = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # Compute optical flow
    flow = cv2.calcOpticalFlowFarneback(gray1, gray2, None, **farneback_params)

    # Compute magnitude and angle
    magnitude, angle = cv2.cartToPolar(flow[..., 0], flow[..., 1])

    # Normalize magnitude
    magnitude = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX)

    # Create HSV image
    hsv = np.zeros_like(frame2)
    hsv[..., 1] = 255
    hsv[..., 0] = angle * 180 / np.pi / 2
    hsv[..., 2] = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX)
    flow_rgb = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

    # Overlay optical flow on black background
    black_background = np.zeros_like(frame2)
    optical_flow_overlay = cv2.addWeighted(black_background, 0.5, flow_rgb, 0.5, 0)

    # Write frame to video
    output.write(optical_flow_overlay)

    # Display the optical flow overlay using matplotlib
    plt.imshow(cv2.cvtColor(optical_flow_overlay, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title(f'Frame {frame_count}')
    plt.show()

    # Save frames if needed (optional)
    plt.savefig(f'frame_{frame_count}.png')
    frame_count += 1

    # Update previous frame
    gray1 = gray2.copy()

cap.release()
output.release()

print("Optical flow dance video and frames saved successfully!")
```
